```python
from litellm import completion
```

```python
response = completion(
    model ="gemini/gemini-3.8-flash",
    messages=[{"role":"user","content":"What is LLM Gateway?"}],
    fallbacks =[        
        "groq:openai/gpt-oss-20b",

    ]
)
```

```python
from langchain_litellm import ChatLiteLLM
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
llm = ChatLiteLLM(model="groq/openai/gpt-oss-20b",temperature=0.3)
```

```python
prompt = ChatPromptTemplate.from_messages([
    ("system","You are a helpful AI tutor named Binsha.be consise"),
    ("user","{question}")
])
```

```python
chain = prompt | llm | StrOutputParser()
answer = chain.invoke({"question":"What is an LLM Gateway ,3 important points"})
print(answer)
```

```python
primary = ChatLiteLLM(model="gemini/gemini-3.8-flash",temperature=0.3)
```

```python
fallback_1 = ChatLiteLLM(model="groq:openai/gpt-oss-20b",temperature=0.3)
fallback_2 = ChatLiteLLM(model="gpt-4o-mini",temperature=0.3)
```

```python
robust_llm = primary.with_fallbacks([fallback_1,fallback_2])
```

```python
prompt = ChatPromptTemplate.from_messages([
    ("system","You are a helpful AI tutor named Binsha.be consise"),
    ("user","{question}")
])
```

```python
chain = prompt | robust_llm | StrOutputParser()
result = chain.invoke({"question":"What are main benefit of LLM Gateway ? "})
result
```
